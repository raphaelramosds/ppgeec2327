Ao todo são M=100 vetores $x_{5,1}$ de entrada \- os quais foram dispostos horizontalmente na figura acima em um grid 10x10 \- cada um com N=5 linhas (features) que podem assumir valor 1 (branco) ou 0 (cinza). Cada vetor $x_{5,1}$ do conjunto de dados é rotulado como 1 ou 0. Então, temos um problema de classificação binária.

Na classificação, a Rede Neural (*Neural Network \- NN)* recebe um vetor de N=5 elementos e clasifica\-o 0 ou 1. Sobre sua estrutura, ela pode ser configurada para ser treinada com $n$ camadas ocultas, cada uma podendo alocar $k^{\left\lbrack i\right\rbrack }$ nós (*nodes*), em que $i$ é o número da camada.

Então, se o programa NNode.m recebeu nodes = \[5 1\] como entrada, temos na verdade uma estrutura \[5 5 1\], em que a primeira posição representa o vetor de entrada de $k^{\left\lbrack 0\right\rbrack } =5$ nós/elementos; logo após, a camada oculta com $k^{\left\lbrack 1\right\rbrack } =5$ nós; e, por fim, a camada de saída, com 1 nó.

# <span style="color:rgb(213,80,0)">Classificação binária (NNode\_test.m)</span>

O gráfico da primeira linha mostra a evolução do erro  *Root Mean Square* (RMS), onde para um conjunto de $N\;$ dados de treino e diferenças $\epsilon_i =t_{\mathrm{obs}} -t_{\mathrm{pred}}$, temos:

$$\mathrm{RMS}=\sqrt{\frac{1}{M}\sum_{i=1}^M \epsilon_i^2 }$$

Os *plots* **b)** e **c)** mostram o estado dos gradientes calculados na iteração 200 (última iteração do treinamento), em que: a primeira camada mapeia 6 nós de entrada (5 features + 1 *bias*) para cada um dos 10 nós da segunda camada; e a segunda camada mapeia 11 nós da camada oculta (10 nós + 1 *bias*) para 1 nó da camada de saída. Matricialmente, portanto, os pesos dessa rede neural podem ser representados por uma matriz

$$\bm{W}=\left\lbrack \begin{array}{cc} {\bm{W}}_{10\times 6}^{\left\lbrack 0\right\rbrack }  & {\bm{W}}_{1\times 11}^{\left\lbrack 1\right\rbrack }  \end{array}\right\rbrack$$

<p style="text-align:left">
   <img src="imagens/NNode_test-error-gradient.jpg" width="445" alt="image_5.jpeg"><img src="imagens/NNode_test-gradient.jpg" width="449" alt="image_6.jpeg">
</p>

O erro RMS convergiu assintoticamente para 0 a partir da iteração 100, indicando que a rede conseguiu aprender bem os padrões dos dados de treino.

Na primeira camada, boa parte dos gradientes retropropagados se tornaram pequenos ( $\left.{\approx 1\times 10}^{-10} \right)$, dadas os tons de amarelo esverdeado serem mais predominante, sinalizando que os pesos atingiram um ponto de mínimo e praticamente pararam de atualizar.

Na camada de saída, por outro lado, temos gradientes ligeiramente maiores do que na primeira camada. Isso é justificado pelo *back\-propagation*, que causa atenuação do erro ( $\mathrm{in}$ ) à medida que é multiplicado pelas derivadas da regra da cadeira rumo às camadas mais profundas/inicias.

## ReLU

Os parâmetros de entrada foram: nodes = \[5 1 6\], norma L2 (1) e ReLU como função de ativação (1).

<p style="text-align:left">
   <img src="imagens/NNode_multiclass-516-relu.jpg" width="756" alt="image_10.jpeg">
</p>

O erro médio inicia em torno de 0.85 e sofre uma queda contínua e acentuada. Ao chegar na iteração 100, o erro reduz\-se a valores próximos de zero. Em relação a precisão de inferência, o alinhamento entre as classes reais e as preditas é praticamente perfeito para todas as 6 classes.

Em contraste da função sigmoide, que sofre com a saturação nas extremidades ( $\sigma \left(z\right)\prime$ tende assintoticamente a 0.0 para pontos suficientemente distantes de $z_0 =0\ldotp 0$ ), a função ReLU possui derivada constante e igual a 1 para entradas positivas. Isso elimina a saturação do gradiente, permitindo que os sinais de erro retropropagados continuem fortes e atualizem a rede. Logo, aumentar o número de nós em alguma camada oculta nesta arquitetura é desnecessário.

Apesar disso, ao permitir que esta arquitetura tenha mais camadas, como no experimento a seguir em que a configuração muda apenas para \[5 1 6 6 6\], nota\-se que há necessidade de um maior número de iterações para a convergência, visto que o custo médio ainda se encontra muito próximo a 0.1 na iteração 100, quando comparado ao experimento anterior, o qual chegou perto de 0.1 mais rapidamente.
